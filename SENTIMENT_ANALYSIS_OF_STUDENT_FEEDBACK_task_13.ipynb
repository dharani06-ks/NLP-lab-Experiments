```python
import pandas as pd

# Create a mock student feedback dataset directly
data = {
    "Feedback": [
        "The lectures were excellent and very useful!",
        "This course is outdated and poor.",
        "Neutral feedback with no strong sentiment.",
        "Clearly a great class, but we need more coding problems.",
        "The professor was good, but the material was bad."
    ]
}
df = pd.DataFrame(data)

positive = ["excellent", "good", "useful", "clearly", "great"]
negative = ["poor", "bad", "outdated", "improvement", "problem"]

def sentiment(text):
    text = str(text).lower()

    pos = sum(word in text for word in positive)
    neg = sum(word in text for word in negative)

    if pos > neg:
        return "Positive"
    elif neg > pos:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Feedback"].apply(sentiment)

print(df[["Feedback", "Sentiment"]])
```
