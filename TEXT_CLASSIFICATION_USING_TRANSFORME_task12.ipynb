```python
# Install once:
# pip install transformers torch

from transformers import pipeline

# Zero-shot classifier can select a topic category
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

text = "Laboratory computers need upgrades."

candidate_labels = [
    "Infrastructure",
    "Academics",
    "Hostel",
    "Sports"
]

result = classifier(text, candidate_labels=candidate_labels)

print("Input:", text)
print("BERT Result:", result["labels"][0])
print("Confidence:", round(result["scores"][0], 4))
print("Category:", result["labels"][0])
```
