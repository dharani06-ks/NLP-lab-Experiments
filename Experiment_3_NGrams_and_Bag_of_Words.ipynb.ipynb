```python
text = "placement training is useful"

print("Input Text:")
print(text)
```

```python
words = text.split()

print("Unigrams:")
for word in words:
    print(word)
```

```python
words = text.split()

print("Bigrams:")

for i in range(len(words) - 1):
    print(words[i], words[i + 1])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = ["placement training is useful"]

vectorizer = CountVectorizer()

bow = vectorizer.fit_transform(documents)

print("Words:")
print(vectorizer.get_feature_names_out())

print("\nBag-of-Words:")
print(bow.toarray())
```
