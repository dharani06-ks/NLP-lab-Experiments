```python
text = "Faculty explains concepts clearly and effectively."

print("Original Text:")
print(text)
```

```python
text = text.lower()

print("Lowercase Text:")
print(text)
```

```python
words = text.split()

print("Tokens:")
print(words)
```

```python
stop_words = {"and", "the", "is", "a", "an", "of", "to", "in"}

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("After Stop-word Removal:")
print(filtered_words)
```

```python
normalized_text =" ".join(filtered_words)

print("Normalized Text:")
print(normalized_text)
```
