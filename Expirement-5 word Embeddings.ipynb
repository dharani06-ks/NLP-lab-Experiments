```python
!pip install gensim
from gensim.models import Word2Vec
sentences=[
    ["faculty","teacher","students"],
    ["placement","recruitment","job"],
    ["hostel","accommodation","room"]
]
#train word2vec model
model=Word2Vec(
sentences,
vector_size=50,
window=2,
min_count=1,
workers=1
)
print("Words similar to faculty:")
print(model.wv.most_similar("faculty"))
print("\nWords similar to placement:")
print(model.wv.most_similar("placement"))
print("\nWords similar to hostel:")
print(model.wv.most_similar("hostel"))
```
